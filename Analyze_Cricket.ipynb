```python
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
import numpy as np # Ensure numpy is imported for np.nan

# --- Start of fix: Ensure match_data_features_final has Match_ID, Winning_Team, and Target ---

# 1. Create a consistent unique Match_ID for both df_bowling and match_data_features_final
# This handles cases where Team1/Team2 order might differ from Team/Vs order in original dataframes
def create_sorted_match_id(row, team1_col, team2_col):
    teams = sorted([row[team1_col], row[team2_col]])
    return f"{row['Date']}_{teams[0]}_vs_{teams[1]}"

# Apply to df_bowling to get a unique match identifier
df_bowling['Match_ID_Unique'] = df_bowling.apply(lambda row: create_sorted_match_id(row, 'Team', 'Vs'), axis=1)

# Get the winning team for each unique match from df_bowling
# We filter for rows where 'Win' is 1 and then get the 'Team' (which is the winning team)
winning_teams_per_match = df_bowling[df_bowling['Win'] == 1].groupby('Match_ID_Unique')['Team'].first().rename('Winning_Team').reset_index()

# Apply the same unique match identifier to match_data_features_final
match_data_features_final['Match_ID_Unique'] = match_data_features_final.apply(lambda row: create_sorted_match_id(row, 'Team1', 'Team2'), axis=1)

# Merge the winning team information into match_data_features_final
match_data_features_final = pd.merge(match_data_features_final, winning_teams_per_match, on='Match_ID_Unique', how='left')

# Define 'Target' column: 1 if Team1 wins, 0 if Team2 wins, np.nan if outcome is unclear
match_data_features_final['Target'] = match_data_features_final.apply(
    lambda row: 1 if row['Team1'] == row['Winning_Team'] else (0 if row['Team2'] == row['Winning_Team'] else np.nan),
    axis=1
)

# Drop rows where 'Target' is NaN (i.e., matches where we couldn't determine a clear winner based on available data)
match_data_features_final.dropna(subset=['Target'], inplace=True)
match_data_features_final['Target'] = match_data_features_final['Target'].astype(int)

# --- End of fix ---

# Define columns to drop for X_final.
# This list now correctly reflects the columns present in match_data_features_final
# and the intent to remove non-predictive identifiers and raw sum features
# that are better represented by derived features like strike rate.
features_to_drop_for_X = [
    'Date',               # Non-predictive unique identifier
    'Match_ID_Unique',    # Non-predictive unique identifier
    'Team1',              # Categorical identifier, not directly a numerical feature for model
    'Team2',              # Categorical identifier
    'Winning_Team',       # Direct leakage of the target variable
    'Target',             # This is the target variable (y_final), not a feature for X
    'Team1_Total_Runs',   # Raw sum, replaced by strike rate
    'Team2_Total_Runs',   # Raw sum, replaced by strike rate
    'Team1_Total_Balls_Faced', # Raw sum, replaced by strike rate
    'Team2_Total_Balls_Faced', # Raw sum, replaced by strike rate
]

X_final = match_data_features_final.drop(columns=features_to_drop_for_X)
y_final = match_data_features_final['Target']

# Handle potential NaN values by filling with the mean of each column
X_final = X_final.fillna(X_final.mean(numeric_only=True))

# Split the data into training and testing sets
X_train_final, X_test_final, y_train_final, y_test_final = train_test_split(X_final, y_final, test_size=0.2, random_state=42)

print(f"\nFinal Training features shape: {X_train_final.shape}")
print(f"Final Testing features shape: {X_test_final.shape}")

# Re-initialize and re-train the RandomForestClassifier model
model_final = RandomForestClassifier(n_estimators=100, random_state=42)
model_final.fit(X_train_final, y_train_final)

print("\nModel re-trained with all specified features.")
```

```python
team1_final_name_predict = 'India'
team2_final_name_predict = 'South africa'

# Extract the latest statistics for the two teams from updated_team_stats
team1_final_stats_predict_final = updated_team_stats[updated_team_stats['Team'] == team1_final_name_predict].drop(columns=['Team']).add_prefix('Team1_')
team2_final_stats_predict_final = updated_team_stats[updated_team_stats['Team'] == team2_final_name_predict].drop(columns=['Team']).add_prefix('Team2_')

# Create a single row DataFrame for this hypothetical final match
hypothetical_final_match_predict_final = pd.concat([team1_final_stats_predict_final.reset_index(drop=True), team2_final_stats_predict_final.reset_index(drop=True)], axis=1)

# Ensure the columns in the hypothetical match data match the order of the training data
hypothetical_final_match_predict_final = hypothetical_final_match_predict_final[X_final.columns]

# Make the prediction using the newly trained model
predicted_final_winner_encoded_predict_final = model_final.predict(hypothetical_final_match_predict_final)

print("\nPrediction for India vs. South Africa (with all requested features including losses):")
if predicted_final_winner_encoded_predict_final[0] == 1:
    print(f"Predicted Winner of the Final Match ({team1_final_name_predict} vs. {team2_final_name_predict}): {team1_final_name_predict}")
else:
    print(f"Predicted Winner of the Final Match ({team1_final_name_predict} vs. {team2_final_name_predict}): {team2_final_name_predict}")

display(hypothetical_final_match_predict_final)
```

```python
# 1. Aggregate all necessary batting statistics for each player
player_batting_stats = df.groupby('Player').agg(
    Total_Runs=('R', 'sum'),
    Total_Balls_Faced=('B', 'sum'),
    Total_4s=('4s', 'sum'),
    Total_6s=('6s', 'sum')
).reset_index()

# 2. Filter players who have faced at least 50 balls for meaningful strike rate calculation
qualified_players = player_batting_stats[player_batting_stats['Total_Balls_Faced'] >= 50].copy()

# 3. Calculate Strike Rate
qualified_players['Strike_Rate'] = qualified_players.apply(
    lambda row: (row['Total_Runs'] / row['Total_Balls_Faced']) * 100 if row['Total_Balls_Faced'] > 0 else 0,
    axis=1
)

# 4. Define and calculate Player Performance Score
# A weighted formula can be used. Weights are illustrative and can be adjusted.
# Example weights: runs have direct impact, strike rate scaled, 6s weighted higher than 4s.
qualified_players['Player_Performance_Score'] = (
    qualified_players['Total_Runs'] +
    (qualified_players['Strike_Rate'] / 5) +
    qualified_players['Total_4s'] +
    (qualified_players['Total_6s'] * 1.5)
)

# 5. Sort by Player_Performance_Score in descending order and display the top 20
top_20_batsmen = qualified_players.sort_values(by='Player_Performance_Score', ascending=False).head(20)

print("Top 20 Batsmen based on Player Performance Score:")
display(top_20_batsmen)
```

```python
import pandas as pd
import numpy as np

# 1. Filter Indian Players from the Batting dataset
india_batting = df[df['Team'] == 'India'].copy()
india_batting['R'] = pd.to_numeric(india_batting['R'], errors='coerce').fillna(0)
india_batting['B'] = pd.to_numeric(india_batting['B'], errors='coerce').fillna(0)
india_batting['4s'] = pd.to_numeric(india_batting['4s'], errors='coerce').fillna(0)
india_batting['6s'] = pd.to_numeric(india_batting['6s'], errors='coerce').fillna(0)

player_batting_summary = india_batting.groupby('Player').agg(
    Matches_Batted=('Date', 'count'),
    Total_Runs=('R', 'sum'),
    Total_Balls_Faced=('B', 'sum'),
    Total_4s=('4s', 'sum'),
    Total_6s=('6s', 'sum')
).reset_index()

player_batting_summary['Batting_Strike_Rate'] = player_batting_summary.apply(
    lambda row: (row['Total_Runs'] / row['Total_Balls_Faced'] * 100) if row['Total_Balls_Faced'] > 0 else 0,
    axis=1
)

# 2. Filter Indian Bowlers from the Bowling dataset
india_bowling = df_bowling[df_bowling['Team'] == 'India'].copy()
india_bowling['W'] = pd.to_numeric(india_bowling['W'], errors='coerce').fillna(0)
india_bowling['R'] = pd.to_numeric(india_bowling['R'], errors='coerce').fillna(0)
india_bowling['O'] = pd.to_numeric(india_bowling['O'], errors='coerce').fillna(0)

player_bowling_summary = india_bowling.groupby('Bowler').agg(
    Matches_Bowled=('Date', 'count'),
    Total_Wickets=('W', 'sum'),
    Total_Runs_Conceded=('R', 'sum'),
    Total_Overs=('O', 'sum')
).reset_index()

player_bowling_summary['Balls_Bowled'] = (player_bowling_summary['Total_Overs'] // 1) * 6 + (player_bowling_summary['Total_Overs'] % 1 * 10).astype(int)
player_bowling_summary['Economy_Rate'] = player_bowling_summary.apply(
    lambda row: (row['Total_Runs_Conceded'] / row['Balls_Bowled'] * 6) if row['Balls_Bowled'] > 0 else 0,
    axis=1
)

# Rename column for merging
player_bowling_summary = player_bowling_summary.rename(columns={'Bowler': 'Player'})

# 3. Combine both summaries
india_squad = pd.merge(player_batting_summary, player_bowling_summary, on='Player', how='outer').fillna(0)

print("--- India Squad Overall Statistics ---")
display(india_squad.sort_values(by=['Total_Runs', 'Total_Wickets'], ascending=False))
```

```python
import re

def clean_player(name):
    """Remove (c), (wk), ?, † marks from player names."""
    name = re.sub(r"\s*\(c\)|\(wk\)|\?|†", "", str(name))
    return " ".join(name.split())

ind_bat  = df[df["Team"].str.strip().str.lower() == "india"].copy()
ind_bowl = df_bowling[df_bowling["Team"].str.strip().str.lower() == "india"].copy()

ind_bat["Player"]  = ind_bat["Player"].map(clean_player)
ind_bowl["Bowler"] = (ind_bowl["Bowler"].map(clean_player)
                      .replace({"V. Chakaravarthy": "Varun Chakravarthy"}))
ind_bowl = ind_bowl[ind_bowl["Bowler"].str.upper() != "BOWLING"]   # stray header row

for c in ["R", "B", "4s", "6s", "SR"]:
    ind_bat[c] = pd.to_numeric(ind_bat[c], errors="coerce")
for c in ["O", "R", "W", "Econ", "Dots", "4s", "6s", "Wide", "NB"]:
    ind_bowl[c] = pd.to_numeric(ind_bowl[c], errors="coerce")

def overs_to_balls(o):                      # 3.4 overs -> 22 legal balls
    if pd.isna(o):
        return 0
    ov = int(o)
    return ov * 6 + int(round((float(o) - ov) * 10))

ind_bowl["Balls"]   = ind_bowl["O"].apply(overs_to_balls)
ind_bat["DateKey"]  = pd.to_datetime(ind_bat["Date"],  format="%d.%m.%y", errors="coerce")
ind_bowl["DateKey"] = pd.to_datetime(ind_bowl["Date"], format="%d.%m.%y", errors="coerce")

# Data-quality fix: the 15.02.26 "vs Pakistan" batting block in the sheet is a
# copy-paste of the 7.2.26 "vs USA" block (dismissals even name USA bowlers).
# Detect any innings whose (R, B, Dismissal) fingerprint repeats an earlier one
# and drop the later copy, so totals are not double counted.
fp    = {k: tuple(sorted(zip(g["R"], g["B"], g["Dismissal"])))
         for k, g in ind_bat.groupby(["Date", "Vs"])}
first = ind_bat.groupby(["Date", "Vs"])["DateKey"].min()
seen, duplicates = {}, []
for key in sorted(fp, key=lambda k: first[k]):
    if fp[key] in seen:
        duplicates.append(key)
    else:
        seen[fp[key]] = key
if duplicates:
    print("[data-quality] dropping copy-pasted innings:", duplicates)
    ind_bat = ind_bat[~ind_bat.set_index(["Date", "Vs"]).index.isin(duplicates)].copy()

print(f"India matches kept for batting: {ind_bat[['Date','Vs']].drop_duplicates().shape[0]}, "
      f"bowling rows: {len(ind_bowl)}")

# %% [markdown]
# ## Cell 2 — Batting aggregates per player

# --- Cell 2: batting scorecard ---
forms = {p: " | ".join(f"{r.R}({r.B})" for r in g.itertuples())
         for p, g in ind_bat.sort_values("DateKey").groupby("Player")}
ducks = (ind_bat.assign(d=(ind_bat["R"] == 0) & (ind_bat["B"] <= 3))
         .groupby("Player")["d"].sum())

bat_agg = (ind_bat.groupby("Player")
           .agg(Innings=("R", "size"), Total_Runs=("R", "sum"), Total_Balls=("B", "sum"),
                Total_4s=("4s", "sum"), Total_6s=("6s", "sum"), Highest_Score=("R", "max"),
                Times_Out=("Dismissal", lambda s: (s != "not out").sum()))
           .reset_index())
bat_agg["Strike_Rate"] = (bat_agg["Total_Runs"] / bat_agg["Total_Balls"].replace(0, pd.NA) * 100).round(1)
bat_agg["Average"]     = (bat_agg["Total_Runs"] / bat_agg["Times_Out"].replace(0, pd.NA)).round(1)
bat_agg["Ducks"]       = bat_agg["Player"].map(ducks).fillna(0).astype(int)
bat_agg["Form"]        = bat_agg["Player"].map(forms)
bat_agg = bat_agg.sort_values("Total_Runs", ascending=False).reset_index(drop=True)
print("INDIA BATTING (chronological form: runs(balls) per match)")
display(bat_agg)

# %% [markdown]
# ## Cell 3 — Bowling aggregates per player

# --- Cell 3: bowling scorecard ---
spells = {p: " | ".join(f"{int(r.W)}/{int(r.R)}" for r in g.itertuples())
          for p, g in ind_bowl.sort_values("DateKey").groupby("Bowler")}
last2  = (ind_bowl.sort_values("DateKey").groupby("Bowler").tail(2)
          .groupby("Bowler").apply(lambda d: d["R"].sum() * 6 / d["Balls"].sum(),
                                   include_groups=False).rename("Last2_Econ"))

bowl_agg = (ind_bowl.groupby("Bowler")
            .agg(Matches=("W", "size"), Balls=("Balls", "sum"), Runs_Given=("R", "sum"),
                 Wickets=("W", "sum"), Dots=("Dots", "sum"),
                 Sixes_Given=("6s", "sum"), Wides=("Wide", "sum"))
            .reset_index().rename(columns={"Bowler": "Player"}))
bowl_agg["Overs"]       = bowl_agg["Balls"].apply(lambda b: f"{b // 6}.{b % 6}")
bowl_agg["Economy"]     = (bowl_agg["Runs_Given"] * 6 / bowl_agg["Balls"]).round(2)
bowl_agg["Dot_Pct"]     = (bowl_agg["Dots"] / bowl_agg["Balls"] * 100).round(1)
bowl_agg["Last2_Econ"]  = bowl_agg["Player"].map(last2).round(2)
bowl_agg["Spells"]      = bowl_agg["Player"].map(spells)
bowl_agg = bowl_agg.sort_values(["Wickets", "Economy"], ascending=[False, True]).reset_index(drop=True)
print("INDIA BOWLING (spell form: wkts/runs per match)")
display(bowl_agg)

# %% [markdown]
# ## Cell 4 — Keep / Remove verdict engine (with reasons)

# --- Cell 4: selection verdicts ---
# Policy thresholds (edit freely and re-run):
BAT_WEAK_SR, BAT_WEAK_AVG   = 110.0, 15.0   # batting weak if SR < 110 and avg < 15
BOWL_WEAK_ECON              = 8.5           # bowling weak if 0 wkts and econ > 8.5
BOWL_EXPENSIVE_ECON         = 11.0          # econ > 11 = disastrous with the ball
BAT_STRONG_SR, BAT_STRONG_RUNS = 130.0, 100 # batting strong if SR >= 130 or runs >= 100
CORE_BOWL_ECON, CORE_BOWL_WKTS = 7.5, 5     # core bowler: econ <= 7.5 and wkts >= 5
CORE_BAT_SR,  CORE_BAT_RUNS    = 140.0, 100 # core batter: SR >= 140 and runs >= 100

B = bat_agg.set_index("Player")
W = bowl_agg.set_index("Player")
rows = []
for p in sorted(set(B.index) | set(W.index)):
    b = B.loc[p] if p in B.index else None
    w = W.loc[p] if p in W.index else None
    reason, notes = [], []

    bat_weak   = (b is not None and b["Innings"] >= 2 and b["Total_Balls"] >= 10
                  and b["Strike_Rate"] < BAT_WEAK_SR
                  and (pd.isna(b["Average"]) or b["Average"] < BAT_WEAK_AVG))
    bat_strong = (b is not None and (b["Strike_Rate"] >= BAT_STRONG_SR
                  or b["Total_Runs"] >= BAT_STRONG_RUNS))
    bat_negl   = (b is None or b["Total_Balls"] < 10)
    bowl_weak  = (w is not None and w["Balls"] >= 18 and w["Wickets"] == 0
                  and w["Economy"] > BOWL_WEAK_ECON)
    bowl_expen = (w is not None and w["Balls"] >= 24 and w["Economy"] > BOWL_EXPENSIVE_ECON)
    bowl_negl  = (w is None or w["Balls"] < 18)

    if bat_weak and (bowl_weak or bowl_negl):
        verdict = "REMOVE"
        reason.append(f"batting only {int(b['Total_Runs'])} runs in {int(b['Innings'])} inns "
                      f"at SR {b['Strike_Rate']} (avg {b['Average']}, {int(b['Total_4s'])} fours "
                      f"/ {int(b['Total_6s'])} sixes)")
        reason.append(f"bowling {int(w['Wickets'])} wkts in {w['Overs']} ov at econ {w['Economy']}"
                      if bowl_weak else "no bowling skill to offset the batting")
    elif bat_negl and bowl_weak:
        verdict = "REMOVE"
        reason.append(f"bowling {w['Overs']} ov, {int(w['Wickets'])} wkts at econ {w['Economy']}, "
                      f"with no batting returns")
    elif bat_strong and bowl_expen:
        verdict = "KEEP (role change)"
        reason.append(f"keep as batter only ({int(b['Total_Runs'])} runs at SR {b['Strike_Rate']}); "
                      f"withdraw the ball - econ {w['Economy']}, "
                      f"{int(w['Sixes_Given'])} sixes given")
    else:
        verdict = "KEEP"
        if w is not None and w["Wickets"] >= CORE_BOWL_WKTS and w["Economy"] <= CORE_BOWL_ECON:
            reason.append(f"core bowler: {int(w['Wickets'])} wkts @ econ {w['Economy']}, "
                          f"{w['Dot_Pct']}% dots")
        if b is not None and b["Total_Runs"] >= CORE_BAT_RUNS and b["Strike_Rate"] >= CORE_BAT_SR:
            reason.append(f"core batter: {int(b['Total_Runs'])} runs @ SR {b['Strike_Rate']}")
        elif b is not None and b["Total_Runs"] >= 40 and b["Strike_Rate"] >= 180:
            reason.append(f"strike-rate outlier (SR {b['Strike_Rate']}) - promote up the order")
        if not reason:
            if b is not None and b["Total_Balls"] >= 20:
                reason.append(f"batting {int(b['Total_Runs'])} runs @ SR {b['Strike_Rate']}")
            if w is not None:
                reason.append(f"bowling {int(w['Wickets'])} wkts @ econ {w['Economy']} "
                              f"in {w['Overs']} ov")

    # management notes
    if (b is not None and w is not None and b["Total_Balls"] >= 30 and w["Balls"] >= 30
            and 8.0 <= w["Economy"] < BOWL_EXPENSIVE_ECON):
        notes.append(f"cap at 2-3 overs in favourable match-ups (overall econ {w['Economy']})")
    if w is not None and w["Balls"] >= 60 and pd.notna(w["Last2_Econ"]) and w["Last2_Econ"] > 10:
        notes.append(f"form dip: last 2 spells at econ {w['Last2_Econ']} - avoid death overs")
    if b is not None and b["Total_Balls"] >= 20 and b["Ducks"] >= 2:
        notes.append(f"volatile ({int(b['Ducks'])} near-golden ducks) - short leash")
    if w is not None and w["Matches"] == 1 and w["Balls"] >= 18 and w["Economy"] <= 7.5 and w["Wickets"] >= 1:
        notes.append("small sample but impressed - rotation option")
    if bat_weak and not bowl_weak and not bowl_negl:
        notes.append("batting is tail-end bonus, not a selection factor")

    rows.append({"Player": p, "Verdict": verdict,
                 "Reason": "; ".join(reason), "Notes": "; ".join(notes)})

order = {"REMOVE": 0, "KEEP (role change)": 1, "KEEP": 2}
verdict_df = (pd.DataFrame(rows)
              .sort_values(["Verdict", "Player"],
                           key=lambda s: s.map(order) if s.name == "Verdict" else s)
              .reset_index(drop=True))
display(verdict_df)

# %% [markdown]
# ## Cell 5 — Final answer: who goes, who stays

# --- Cell 5: plain-language summary ---
print("=" * 78)
print("PLAYERS TO REMOVE FROM TEAM INDIA")
print("=" * 78)
for r in verdict_df[verdict_df["Verdict"] == "REMOVE"].itertuples():
    print(f"\n* {r.Player}\n    Reason: {r.Reason}")

print("\n" + "=" * 78)
print("PLAYERS WHO SHOULD STAY")
print("=" * 78)
for r in verdict_df[verdict_df["Verdict"] != "REMOVE"].itertuples():
    line = f"\n* {r.Player}  [{r.Verdict}]\n    Reason: {r.Reason}"
    if r.Notes:
        line += f"\n    Note:   {r.Notes}"
    print(line)

# optional: download the verdict table from Colab
verdict_df.to_csv("india_player_verdicts.csv", index=False)
try:
    from google.colab import files
    files.download("india_player_verdicts.csv")
except Exception:
    print("\n[saved] india_player_verdicts.csv")
```
